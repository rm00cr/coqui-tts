```python
import pandas as pd
import os
import shutil
from pathlib import Path
from tqdm import tqdm
import subprocess

# Your selected speakers
good_audios_DE = ["419101","412497","412827","412831","413194","413570","414560","414863","415138","415624"]
good_audios_IT = ["413028","415812","415909","416492","416531","416773","416873","417448","418256","418774"]
good_audios_FR = ["412440","412522","413217","413330","413579","414037","414792","414843","414927","414992"]
good_audios_ES = ["412907","413233","414394","414554","414661","414852","415738","415878","418171","418189"]
```

```python

def extract_speaker_audios(data_dir, output_dir, lan):
    """Extract audio files for selected speakers and organize by speaker ID and language"""
    
    data_path = Path(data_dir)
    output_path = Path(output_dir)
    output_path.mkdir(exist_ok=True)
    
    # Language mapping
    languages = {
        'DE': good_audios_DE,
        'IT': good_audios_IT, 
        'FR': good_audios_FR,
        'ES': good_audios_ES
    }
    good_audios = languages[lan]
    
    
    folders = os.listdir(data_dir+"/"+lan+"/audios/")
    print(f"Found {len(folders)} audio files")


    for speaker_id in good_audios:
        speaker_dir = f"{output_dir}/{lan}_{speaker_id}/"
        os.makedirs(speaker_dir, exist_ok=True)
    
    copied_files = {speaker_id: 0 for speaker_id in good_audios}

    for file in tqdm(folders, desc=f"Processing {lan} files"):
        for i,speaker_id in enumerate(good_audios):
            if speaker_id in file:
                # Create output path
                opath = f"{output_dir}/{lan}_{speaker_id}/"
                
                # Input file path
                input_file = f"{data_dir}/{lan}/audios/{file}"
                
                # Determine output filename (convert to .wav)
                if file.endswith('.mp3'):
                    output_filename = file.replace('.mp3', '.wav')
                else:
                    output_filename = file.replace('.wav', '.wav')  # Keep .wav
                
                output_file = f"{opath}{output_filename}"
                
                # Check if output file already exists
                if os.path.exists(output_file):
                    copied_files[speaker_id] += 1
                    continue
                
                try:
                    if file.endswith('.mp3'):
                        # Convert MP3 to WAV using ffmpeg
                        cmd = [
                            'ffmpeg', '-i', input_file,
                            '-ar', '24000',  # 24kHz sample rate
                            '-ac', '1',      # mono
                            '-y',            # overwrite
                            output_file
                        ]
                        subprocess.run(cmd, capture_output=True, check=True)
                    else:
                        # Just copy if already WAV
                        shutil.copy2(input_file, output_file)
                    
                    copied_files[speaker_id] += 1
                    
                except Exception as e:
                    print(f"Error processing {file}: {e}")
                
                break  # Found the speaker, no need to check others
             
    # Print summary
    print(f"/nSummary for {lan}:")
    for speaker_id, count in copied_files.items():
        print(f"  Speaker {speaker_id}: {count} files")
```

```python
data_dir = "/home/romolo/VT1/data"
output_dir = "/home/romolo/VT1/selected_speakers"

# Process each language
for language in ['DE', 'IT', 'FR', 'ES']:
    extract_speaker_audios(data_dir, output_dir, language)
```
