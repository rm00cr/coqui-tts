```python
def get_language_from_dataset(dataset_name):
    """Extract language code from dataset name.
    Example: 'mls_english_dev' -> 'en'
    """
    language_map = {
        'english': 'en',
        'spanish': 'es',
        'french': 'fr',
        'german': 'de',
        'italian': 'it',
        'portuguese': 'pt',
        'polish': 'pl',
        'dutch': 'nl',
    }
    
    dataset_lower = dataset_name.lower()
    for lang_name, lang_code in language_map.items():
        if lang_name in dataset_lower:
            return lang_code
    
    return 'en'
df['lang'] = df['dataset'].apply(get_language_from_dataset)
#df.to_csv('/data/train/metadata/metadata/result/100speaker_perlang.csv')
```
